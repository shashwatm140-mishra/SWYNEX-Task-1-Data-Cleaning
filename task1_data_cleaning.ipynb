```python
import pandas as pd
import numpy as np

df = pd.read_csv("titanic.csv")
df.head()
```

```python
# Check the size of the dataset
print("Dataset shape:", df.shape)

# Check column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types
print("\nData types:")
print(df.dtypes)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

## 2. Initial Findings

The dataset contains 891 rows and 12 columns.

The initial inspection identified missing values in three columns:

- Age: 177 missing values
- Cabin: 687 missing values
- Embarked: 2 missing values

No duplicate rows were identified.

Most columns have appropriate data types. Numerical variables such as Age and Fare are stored as numeric types, while categorical/text variables such as Sex, Cabin, and Embarked are stored as object types.

### Cleaning Plan

- Age: Missing values will be replaced using the median age because Age is a numerical variable and the median is less affected by extreme values.
- Cabin: The column has a very high proportion of missing values. Instead of inventing cabin numbers, missing values will be represented as "Unknown".
- Embarked: The two missing values will be filled using the mode because Embarked is a categorical variable.
- Duplicate rows: No duplicates were found, so no rows will be removed.

```python
# Calculate the median age
median_age = df["Age"].median()

print("Median age:", median_age)
```

```python
# Fill missing Age values with the median age
df["Age"] = df["Age"].fillna(median_age)

# Verify that Age has no missing values
print("Missing Age values after cleaning:", df["Age"].isnull().sum())
```

```python
# Replace missing Cabin values with "Unknown"
df["Cabin"] = df["Cabin"].fillna("Unknown")

# Verify that Cabin has no missing values
print("Missing Cabin values after cleaning:", df["Cabin"].isnull().sum())
```

```python
# Find the most frequent Embarked value
embarked_mode = df["Embarked"].mode()[0]

print("Most frequent Embarked value:", embarked_mode)
```

```python
# Fill missing Embarked values with the mode
df["Embarked"] = df["Embarked"].fillna(embarked_mode)

# Verify that Embarked has no missing values
print("Missing Embarked values after cleaning:", df["Embarked"].isnull().sum())
```

```python
# Check missing values after cleaning
print("Missing values after cleaning:")
print(df.isnull().sum())

# Check duplicate rows again
print("\nDuplicate rows after cleaning:", df.duplicated().sum())
```

```python
# Check data types after cleaning
print("Data types after cleaning:")
print(df.dtypes)
```

```python
# Check numerical columns for unusual values

print("Age range:", df["Age"].min(), "to", df["Age"].max())
print("Fare range:", df["Fare"].min(), "to", df["Fare"].max())
print("Pclass values:", sorted(df["Pclass"].unique()))
print("Survived values:", sorted(df["Survived"].unique()))
print("SibSp range:", df["SibSp"].min(), "to", df["SibSp"].max())
print("Parch range:", df["Parch"].min(), "to", df["Parch"].max())
```

```python
# Final validation of the cleaned dataset

print("Dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows of cleaned data:")
display(df.head())
```

```python
# Save the cleaned dataset
df.to_csv("cleaned_titanic.csv", index=False)

print("Cleaned dataset saved successfully as cleaned_titanic.csv")
```

## 3. Data Cleaning Summary

The Titanic dataset was successfully cleaned and prepared for further analysis.

### Cleaning actions performed

| Issue | Action Taken |
|---|---|
| Missing Age values (177) | Replaced with the median age (28.0) |
| Missing Cabin values (687) | Replaced with "Unknown" |
| Missing Embarked values (2) | Replaced with the mode ("S") |
| Duplicate rows | No duplicates were found |
| Data types | Verified as appropriate for analysis |
| Numerical values | Checked for unusual or invalid ranges |

### Assumptions

1. The median was used for missing Age values because Age is numerical and the median is less sensitive to extreme values.
2. Missing Cabin values were represented as "Unknown" rather than assigning fabricated cabin information.
3. Missing Embarked values were replaced with the most frequent category, "S".
4. Existing data types were retained because they were suitable for analysis and future machine-learning applications.
5. No rows were removed because of missing values, as doing so would result in substantial data loss.

The cleaned dataset was saved as `cleaned_titanic.csv`.
